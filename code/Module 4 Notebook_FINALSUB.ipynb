*Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)*

LUAD arises in the peripheral/distal lung, specifically the alveolar and bronchiolar epithelium, in contrast to squamous cell carcinoma, which originates in central airways. The cells lining the alveoli, alveolar type I (AT1) cells (responsible for gas exchange) and alveolar type II (AT2) cells (which produce surfactant and serve as stem-like progenitors), are the primary cells of origin for LUAD.

Under oncogenic stress (e.g., KRAS^G12D^ activation or carcinogen exposure), AT2 cells can dedifferentiate into a KRT8+ alveolar intermediate cell (KAC) state, a plastic, injury-responsive intermediate that has been identified as a key precursor to malignant transformation. LUAD follows a stepwise histological progression:

(1) Atypical adenomatous hyperplasia(AAH) 
(2) Adenocarcinoma in situ (AIS) 
(3) Minimally invasive adenocarcinoma(MIA) 
(4) Invasive adenocarcinoma

Histologically, LUAD growth patterns range from lepidic (low-grade; cells spread along existing alveolar walls) to acinar/papillary (intermediate-grade) to micropapillary/solid (high-grade). The proportion of high-grade components within a tumor is a key prognostic indicator.

### Methods

*The machine learning techniques we are using are: Principal Component Analysis (PCA), UMAP, and K-Means Clustering.* Principal Component Analysis (PCA) is a dimensionality reduction technique that finds the directions of maximum variance in high-dimensional data and projects each sample onto those axes. PCA compresses the immense number of genes per person into a small number of "principal components" that capture the most important axes of transcriptomic variation. PC1 explains the most variance, PC2 the second most, and so on. By plotting PC1 vs. PC2, we get the best possible 2D linear summary of the full gene expression landscape across our 80 LUAD patients.

*What is PCA optimizing?* PCA finds the linear combination of genes that maximizes explained variance. Each successive principal component is orthogonal (uncorrelated) to all previous ones and explains as much remaining variance as possible. It does not use any clinical labels, it is purely driven by the structure of the expression data itself.

*UMAP (Uniform Manifold Approximation and Projection)* is a non-linear dimensionality reduction method that preserves local neighborhood structure: patients with similar overall gene expression profiles end up close together in the 2D projection. Unlike PCA, UMAP can capture curved or complex manifold structure that a linear projection would flatten out. The key hyperparameters are n_neighbors (how many nearby samples define a "neighborhood") and min_dist (how tightly samples are packed in the projection).

*K-Means Clustering* partitions patients into k non-overlapping groups by iteratively assigning each sample to its nearest cluster centroid and updating centroids until assignments stabilize. It is run on the top principal components (rather than all 15,716 raw genes) to reduce noise. The optimal number of clusters is chosen by maximizing the silhouette score, which measures how similar each sample is to its own cluster compared to neighboring clusters, a score of 1.0 means perfectly separated clusters, 0 means clusters are overlapping.

*What is K-Means optimizing?* K-Means minimizes within-cluster sum of squared distances, the total distance between each point and its cluster centroid. It stops when cluster assignments no longer change between iterations, or when a maximum iteration limit is reached.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import umap

import warnings
warnings.filterwarnings('ignore')
```

## Load and subset data

```python
# Load expression matrix (genes x samples)
# For unsupervised EDA (PCA, UMAP, clustering) we use the TRAINING SET only.
# The validation and test sets are reserved strictly for supervised learning evaluation.
expr_raw = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TRAINING_SET_GSE62944_subsample_log2TPM.csv', index_col=0)
print(f'Expression matrix shape (genes x samples): {expr_raw.shape}')

# Load clinical metadata
meta_all = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TRAINING_SET_GSE62944_metadata.csv')
print(f'Metadata shape: {meta_all.shape}')
print(f'Cancer types: {meta_all["cancer_type"].value_counts().to_dict()}')

# Subset to LUAD only
luad_meta = meta_all[meta_all['cancer_type'] == 'LUAD'].copy().reset_index(drop=True)
luad_expr = expr_raw[luad_meta['sample'].values]   # genes x LUAD samples
print(f'LUAD expression shape (genes x samples): {luad_expr.shape}')
print(f'LUAD metadata shape: {luad_meta.shape}')

# Transpose: samples x genes (standard ML format)
X_all_genes = luad_expr.T   # samples x all genes
X_all_genes.index.name = 'sample'

print(f'\nTransposed expression matrix (samples x genes): {X_all_genes.shape}')

# Define the 9-gene metastasis signature (Shen et al. 2019)
SIG_GENES = ['HMMR', 'B4GALT1', 'SLC16A3', 'ANGPTL4', 'EXT1', 'GPC1', 'RBCK1', 'SOD1', 'AGRN']

# Confirm all genes are present; check known aliases if any are missing
ALIASES = {'HMMR': 'RHAMM', 'RBCK1': 'HOIL1L'}
missing = [g for g in SIG_GENES if g not in X_all_genes.columns]
if missing:
    for g in missing:
        alias = ALIASES.get(g)
        if alias and alias in X_all_genes.columns:
            print(f'  Using alias {alias} for {g}')
            X_all_genes = X_all_genes.rename(columns={alias: g})
        else:
            print(f'WARNING — {g} not found (alias: {alias})')
    missing = [g for g in SIG_GENES if g not in X_all_genes.columns]

if not missing:
    print(f'All {len(SIG_GENES)} signature genes confirmed present in dataset.')

# Subset expression to signature genes only
X_sig = X_all_genes[SIG_GENES].copy()
print(f'Signature gene matrix shape: {X_sig.shape}')

# Compute average signature expression score per patient (mean of 9 log2-TPM values)
luad_meta = luad_meta.copy()
luad_meta.index = luad_meta['sample'].values
luad_meta['sig_score'] = X_sig.mean(axis=1).values

print(f'\nSignature score summary:')
print(luad_meta['sig_score'].describe().round(3))
```

## Compute score and clean data

```python
# clinical features for visualization

# Tumor stage: simplify to Stage I / II / III / IV
def simplify_stage(s):
    if pd.isna(s) or '[' in str(s):
        return 'Unknown'
    if 'IV' in s:
        return 'Stage IV'
    if 'III' in s:
        return 'Stage III'
    if 'II' in s:
        return 'Stage II'
    if 'I' in s:
        return 'Stage I'
    return 'Unknown'

# Tumor status: replace [Not Available] and [Unknown] with 'Unknown'
luad_meta['stage_simple'] = luad_meta['ajcc_pathologic_tumor_stage'].apply(simplify_stage)
luad_meta['tumor_status_clean'] = luad_meta['tumor_status'].replace(
    {'[Not Available]': 'Unknown', '[Unknown]': 'Unknown'}
)

# print statements
print('Stage distribution:')
print(luad_meta['stage_simple'].value_counts())
print('\nTumor status distribution:')
print(luad_meta['tumor_status_clean'].value_counts())
```

## Signature Gene Overview

```python
# Visualize distribution of signature gene expression across LUAD patients
fig, axes = plt.subplots(3, 3, figsize=(13, 9))
axes = axes.flatten()

n_patients = len(X_sig)

# Plot histogram for each signature gene
for i, gene in enumerate(SIG_GENES):
    axes[i].hist(X_sig[gene], bins=20, color='steelblue', edgecolor='white', alpha=0.85)
    axes[i].set_title(gene, fontsize=12, fontweight='bold')
    axes[i].set_xlabel('log2 TPM')
    axes[i].set_ylabel('Count')
    mean_val = X_sig[gene].mean()
    axes[i].axvline(mean_val, color='firebrick', linestyle='--', linewidth=1.5, label=f'Mean: {mean_val:.2f}')
    axes[i].legend(fontsize=8)

plt.suptitle(f'Distribution of 9-Gene Metastasis Signature in LUAD (n={n_patients})',
             fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

## PCA: Dimensionality Reduction

We apply PCA to the full gene expression matrix (all 15,716 genes), scaled to zero mean and unit variance. Using all genes captures global transcriptomic variation and places the 9-gene signature in broader biological context. We then color the PCA by:

1. Average signature score:does the metastasis signature drive variance?
2. Tumor stage: does transcriptomic variation correspond to clinical stage?
3. Tumor status: tumor free vs. with tumor

```python
#Scale and run PCA on all genes 
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_all_genes)

# PCA with 20 components (enough to capture most variance, but not too many to overfit)
pca = PCA(n_components=20, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# Variance explained
var_exp = pca.explained_variance_ratio_ * 100
print('Variance explained by top 10 PCs:')
for i, v in enumerate(var_exp[:10]):
    print(f'  PC{i+1}: {v:.2f}%')
print(f'  Total (PC1-5): {var_exp[:5].sum():.2f}%')

#scree plot and cumulative variance plot
# I used Claude to help me design the layout of the figure to effectively communicate the variance explained by each principal component. The side-by-side arrangement allows for easy comparison between the individual variance explained (scree plot) and the cumulative variance explained, which is crucial for determining how many PCs to retain for downstream analysis. The use of distinct colors and clear labeling enhances readability and helps highlight the key insights from the PCA results.
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.bar(range(1, 21), var_exp, color='steelblue', edgecolor='white')
ax1.set_xlabel('Principal Component')
ax1.set_ylabel('Variance Explained (%)')
ax1.set_title('Scree Plot — Variance per PC')
ax1.set_xticks(range(1, 21))
ax2.plot(range(1, 21), np.cumsum(var_exp), 'o-', color='firebrick')
ax2.axhline(80, color='gray', linestyle='--', alpha=0.7, label='80% threshold')
ax2.set_xlabel('Number of PCs')
ax2.set_ylabel('Cumulative Variance Explained (%)')
ax2.set_title('Cumulative Variance Explained')
ax2.legend()
ax2.set_xticks(range(1, 21))

# Overall title and layout adjustments
plt.suptitle('PCA Variance Explained — LUAD (all genes)', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
#PCA colored by signature score, stage, and tumor status
pc1, pc2 = X_pca[:, 0], X_pca[:, 1]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Panel A: Colored by average 9-gene signature score
# I used Claude to help me format and give me template to build the graph off of.
sc = axes[0].scatter(pc1, pc2,
                     c=luad_meta['sig_score'].values,
                     cmap='RdYlBu_r', s=80, alpha=0.85, edgecolors='white', linewidth=0.4)
plt.colorbar(sc, ax=axes[0], label='Avg Signature Score (log2 TPM)')
axes[0].set_xlabel(f'PC1 ({var_exp[0]:.1f}% variance)')
axes[0].set_ylabel(f'PC2 ({var_exp[1]:.1f}% variance)')
axes[0].set_title('A) Colored by Signature Score\n(avg of 9 metastasis genes)')

# Panel B: Colored by pathological tumor stage
# I used Claude to help me format and give me template to build the graph off of.
stage_order = ['Stage I', 'Stage II', 'Stage III', 'Stage IV', 'Unknown']
stage_colors = {'Stage I': '#4daf4a', 'Stage II': '#377eb8',
                'Stage III': '#ff7f00', 'Stage IV': '#e41a1c', 'Unknown': '#999999'}
for stage in stage_order:
    mask = luad_meta['stage_simple'].values == stage
    if mask.sum() > 0:
        axes[1].scatter(pc1[mask], pc2[mask],
                       c=stage_colors[stage], label=stage, s=80, alpha=0.85,
                       edgecolors='white', linewidth=0.4)
axes[1].set_xlabel(f'PC1 ({var_exp[0]:.1f}% variance)')
axes[1].set_ylabel(f'PC2 ({var_exp[1]:.1f}% variance)')
axes[1].set_title('B) Colored by Tumor Stage')
axes[1].legend(loc='best', fontsize=8)

# Panel C: Colored by tumor status (TUMOR FREE / WITH TUMOR / Unknown)
# I used Claude to help me format and give me template to build the graph off of.
status_colors = {'TUMOR FREE': '#2ca02c', 'WITH TUMOR': '#d62728', 'Unknown': '#aec7e8'}
for status, color in status_colors.items():
    mask = luad_meta['tumor_status_clean'].values == status
    if mask.sum() > 0:
        axes[2].scatter(pc1[mask], pc2[mask],
                       c=color, label=f'{status} (n={mask.sum()})',
                       s=80, alpha=0.85, edgecolors='white', linewidth=0.4)
axes[2].set_xlabel(f'PC1 ({var_exp[0]:.1f}% variance)')
axes[2].set_ylabel(f'PC2 ({var_exp[1]:.1f}% variance)')
axes[2].set_title('C) Colored by Tumor Status')
axes[2].legend(loc='best', fontsize=8)

n_patients = len(luad_meta)
n_genes = X_all_genes.shape[1]
plt.suptitle(f'PCA of LUAD Gene Expression (n={n_patients} samples, {n_genes:,} genes)',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## K-Means Clustering on PCA Space

We apply K-Means clustering to the top principal components (capturing the majority of variance) to ask: do LUAD patients naturally separate into transcriptomically distinct subgroups?

We use the silhouette score to select the optimal number of clusters k, then overlay cluster labels on the PCA plot.

```python
# K-Means Clustering: choose optimal k using silhouette score

# Step 1: Determine how many PCs to retain
# We use PCs that collectively explain >= 80% of variance.
# This reduces noise from low-variance components while retaining most biological signal.
cumvar = np.cumsum(var_exp)
n_pcs = np.argmax(cumvar >= 80) + 1
print(f'Using top {n_pcs} PCs (explain {cumvar[n_pcs-1]:.1f}% of variance)')

# Subset the PCA matrix to the retained components
X_pca_reduced = X_pca[:, :n_pcs]

# Step 2: Silhouette score to select optimal k
# Silhouette measures how similar a sample is to its own cluster vs. neighboring clusters.
# Score ranges from -1 (wrong cluster) to +1 (well-separated). We want the highest score.
sil_scores = []
k_range = range(2, 9)  # test k=2 through k=8

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_pca_reduced)
    sil = silhouette_score(X_pca_reduced, labels)
    sil_scores.append(sil)
    print(f'  k={k}: silhouette = {sil:.4f}')

# Step 3: Pick the k with the highest silhouette score
best_k = list(k_range)[np.argmax(sil_scores)]
print(f'\nOptimal k = {best_k} (silhouette = {max(sil_scores):.4f})')

# Step 4: Plot silhouette scores vs. k
# The vertical dashed line marks the best k.
plt.figure(figsize=(7, 4))
plt.plot(list(k_range), sil_scores, 'o-', color='steelblue', linewidth=2, markersize=8)
plt.axvline(best_k, color='firebrick', linestyle='--', label=f'Best k={best_k}')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score vs. k — K-Means on PCA Space')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Fit final K-Means with best k
km_final = KMeans(n_clusters=best_k, random_state=42, n_init=10)
cluster_labels = km_final.fit_predict(X_pca_reduced)
luad_meta['cluster'] = cluster_labels

cluster_palette = sns.color_palette('tab10', best_k)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# I used Claude to make the figures look better and more informative by plotting the K-Means clusters on both the PCA and UMAP visualizations. This allows for a clear comparison of how the clusters are distributed in both dimensionality reduction spaces, providing insights into the underlying structure of the data. The use of distinct colors for each cluster, along with legends and axis labels, enhances interpretability and helps to communicate the key findings from the clustering analysis effectively.
# Clusters on PCA
for c in range(best_k):
    mask = cluster_labels == c
    axes[0].scatter(pc1[mask], pc2[mask],
                   color=cluster_palette[c], label=f'Cluster {c} (n={mask.sum()})',
                   s=90, alpha=0.85, edgecolors='white', linewidth=0.5)
axes[0].set_xlabel(f'PC1 ({var_exp[0]:.1f}% variance)')
axes[0].set_ylabel(f'PC2 ({var_exp[1]:.1f}% variance)')
axes[0].set_title(f'K-Means Clusters (k={best_k}) on PCA')
axes[0].legend(fontsize=9)

#Clusters on UMAP
for c in range(best_k):
    mask = cluster_labels == c
    axes[1].scatter(u1[mask], u2[mask],
                   color=cluster_palette[c], label=f'Cluster {c}',
                   s=90, alpha=0.85, edgecolors='white', linewidth=0.5)
axes[1].set_xlabel('UMAP 1'); axes[1].set_ylabel('UMAP 2')
axes[1].set_title(f'K-Means Clusters (k={best_k}) on UMAP')
axes[1].legend(fontsize=9)

plt.suptitle(f'K-Means Clustering of LUAD Patients (k={best_k})',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

# characterize clusters by signature score and stage 
# I used Claude to do the string cacatenation
print('=== Cluster Characterization ===')
print('\nMean signature score by cluster:')
print(luad_meta.groupby('cluster')['sig_score'].agg(['mean','std','count']).round(3))

print('\nTumor stage distribution by cluster:')
print(pd.crosstab(luad_meta['cluster'], luad_meta['stage_simple']))

print('\nTumor status distribution by cluster:')
print(pd.crosstab(luad_meta['cluster'], luad_meta['tumor_status_clean']))
```

The absence of discrete UMAP clusters suggests that LUAD patients in this dataset *do not fall into transcriptomically distinct subtypes*, instead, they vary continuously. The silhouette score was maximized at *k=2*, but Cluster 1 contains only ~3 patients, the same outliers flagged by PCA and UMAP. *Connection to our project goal:* The unsupervised analysis tells us there are no obvious natural groupings in LUAD transcriptomics that would make classification trivial. The 9-gene signature must provide specific, localized signal that is not visible in the first two principal components.

### Supervised Learning Plan & Expectations

Based on the unsupervised results, we expect our supervised model to face a *challenging classification problem*:

1. *No obvious linear separability*: PCA shows no clear decision boundary by tumor status in global expression space. This suggests a linear model may struggle, but we will test it anyway because regularization can still extract weak signal from a noisy, high-dimensional-relative-to-sample setting.

2. *Expected performance*: Given the lack of unsupervised separation, we expect validation AUC to be *above 0.50 (better than random) but well below 0.80*, likely in the 0.60–0.75 range. The signature was validated for survival prediction in a larger cohort; our small sample and surrogate outcome (tumor status) will attenuate performance.

3. *Model choice*: We will use a *Random Forest Classifier* as our baseline because it is robust to the small, imbalanced cohort (only ~17 WITH TUMOR vs. ~45 TUMOR FREE), uses *class_weight='balanced'* to prevent majority-class bias, and produces interpretable feature importances. If Random Forest overfits (Train AUC >> Val AUC), we will switch to a *Logistic Regression with L2 regularization*, which constrains coefficient magnitude and generalizes better on small samples.

4. *Primary metric*: Cross-validated *AUC (Area Under the ROC Curve)*, because it correctly handles class imbalance and evaluates discrimination across all thresholds — a single accuracy number would be misleading given the 3:1 class ratio.


```python
# Supervised learning: initial cross-validated Random Forest on the TRAINING set
# (This uses the same LUAD training data loaded in Cell 8 — X_sig and luad_meta)
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import roc_auc_score, RocCurveDisplay, ConfusionMatrixDisplay
import warnings
warnings.filterwarnings('ignore')

# Build classification dataframe from training LUAD samples with known tumor status
clf_df = X_sig.copy()
clf_df.index = luad_meta['sample'].values
clf_df['tumor_status'] = luad_meta['tumor_status_clean'].values

clf_df = clf_df[clf_df['tumor_status'].isin(['TUMOR FREE', 'WITH TUMOR'])].copy()

# Features and binary labels (1 = WITH TUMOR, 0 = TUMOR FREE)
X_clf = clf_df[SIG_GENES].values
y_clf = (clf_df['tumor_status'] == 'WITH TUMOR').astype(int).values

print(f"Training LUAD samples: n={len(clf_df)}  |  WITH TUMOR: {y_clf.sum()}  |  TUMOR FREE: {(y_clf==0).sum()}")

# Fit Random Forest on full training set (for feature importance visualization only)
# NOTE: This in-sample fit is only for feature importance — do NOT use these predictions for AUC
rf = RandomForestClassifier(n_estimators=500, class_weight='balanced',
                             random_state=42, max_features='sqrt')
rf.fit(X_clf, y_clf)

# Feature importances
feature_names = SIG_GENES
importances = pd.Series(rf.feature_importances_, index=feature_names).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(8, 6))
colors = ['#d62728' if i == importances.index[-1] else '#1f77b4' for i in importances.index]
importances.plot(kind='barh', ax=ax, color=colors, edgecolor='white')
ax.set_xlabel('Mean Decrease in Impurity (Feature Importance)')
ax.set_title('Random Forest Feature Importances\n9-Gene Signature Predicting Tumor Status (LUAD, Training Set)',
             fontweight='bold')
ax.axvline(1/len(feature_names), color='gray', linestyle='--', alpha=0.7, label='Uniform baseline (1/9)')
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

print("\nFeature importances (ranked):")
print(importances.sort_values(ascending=False).round(4))
```

## Interrogating the Baseline Model: Training Performance & Feature Importances

### Training Dataset Performance

The Random Forest was fit on all LUAD training samples with known tumor status and then evaluated on those *same training samples* to compute the in-sample (training) AUC. The training AUC is *≈ 1.000*, basically perfect.

This is a critical diagnostic result: *a perfect training AUC means the model has memorized the training data, not learned generalizable patterns.* With only ~56 training samples and 500 decision trees, each tree can grow deep enough to create leaf nodes containing single patients. The model is recalling individual training samples by their exact expression profiles, the model is overfitting.

Why does this matter for our project? It tells us that the 9-gene signature, while biologically meaningful, does not contain a strong enough signal to confidently classify tumor status in a small cohort without regularization. The Random Forest is too flexible for this dataset size.

## Evaluating the Model with Cross-Validated AUC

Our primary evaluation metric is *cross-validated AUC*. AUC measures the probability that the model ranks a randomly chosen WITH TUMOR patient higher than a randomly chosen TUMOR FREE patient, across all possible classification thresholds. An AUC of 0.50 is random chance; 1.00 is perfect discrimination. We chose AUC over accuracy because our classes are imbalanced (~17 WITH TUMOR vs. ~45 TUMOR FREE): a model that always predicts TUMOR FREE would achieve ~73% accuracy but an AUC of exactly 0.50, correctly flagging it as uninformative.

We use *5-fold stratified cross-validation* so the model is trained and evaluated on five non-overlapping splits with class ratios preserved. This prevents the training AUC ≈ 1.000 from inflating our estimate of real-world performance. We additionally inspect the *confusion matrix* to confirm the model is actually predicting both classes, not just predicting the majority class on every sample.

*External validation benchmark:* Shen et al. (2019) reported that HMMR and ANGPTL4 were the top prognostic genes in their independent LUAD validation cohort. If our model's top feature importances agree with theirs, that is biological corroboration that our model is capturing real signal, not dataset-specific noise.


```python
# Cross-validated AUC
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Get out-of-fold predicted probabilities
y_prob_cv = cross_val_predict(rf, X_clf, y_clf, cv=skf, method='predict_proba')[:, 1]
auc_cv = roc_auc_score(y_clf, y_prob_cv)
print(f'5-Fold CV AUC: {auc_cv:.4f}')
print('(0.50 = random chance | 1.00 = perfect)')

# Per-fold breakdown
# I used Claude to show me how to create the AUC for each fold of the cross-validation
fold_aucs = []
for fold, (tr, te) in enumerate(skf.split(X_clf, y_clf)):
    rf_fold = RandomForestClassifier(n_estimators=500, class_weight='balanced',
                                      random_state=42, max_features='sqrt')
    rf_fold.fit(X_clf[tr], y_clf[tr])
    prob = rf_fold.predict_proba(X_clf[te])[:, 1]
    auc = roc_auc_score(y_clf[te], prob)
    fold_aucs.append(auc)
    print(f'  Fold {fold+1}: AUC = {auc:.4f}')

# i used claude to string cacatenate for clarity
print(f'\nMean AUC: {np.mean(fold_aucs):.4f} ± {np.std(fold_aucs):.4f}')

# ROC curve for cross-validated predictions
fig, ax = plt.subplots(figsize=(7, 6))
RocCurveDisplay.from_predictions(y_clf, y_prob_cv, ax=ax,
                                  name=f'Random Forest (CV AUC = {auc_cv:.3f})',
                                  color='#d62728')
ax.plot([0,1],[0,1], 'k--', label='Random chance (AUC = 0.50)')
ax.set_title('ROC Curve — 9-Gene Signature Predicting Tumor Status\n(5-Fold CV, LUAD n={})'.format(len(clf_df)),
             fontweight='bold')
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()

# Confusion matrix
# claude was used to format the confusion matrix for better visual appeal and clarity. 
y_pred_cv = (y_prob_cv > 0.5).astype(int)
fig, ax = plt.subplots(figsize=(5, 4))
ConfusionMatrixDisplay.from_predictions(y_clf, y_pred_cv,
                                         display_labels=['TUMOR FREE', 'WITH TUMOR'],
                                         cmap='Blues', ax=ax)
ax.set_title('Confusion Matrix (CV predictions)', fontweight='bold')
plt.tight_layout()
plt.show()
```

## Training & Evaluation: In-Sample vs. Out-of-Sample Error

To properly assess model generalization, we split the LUAD cohort into a *training set*  and a *validation set* using stratified splitting to preserve the class ratios. We fit the Random Forest on the training set only, computed *in-sample error* (performance on the training set), and then evaluated *out-of-sample error* (performance on the held-out validation set). This distinction was critical: as a model that is perfectly accurate on training data but poor on validation data is *overfit*, meaning it has memorized the training samples rather than learning generalizable patterns.

We use *AUC (Area Under the ROC Curve)* as our primary metric because:
- Our classes are imbalanced (≈17 WITH TUMOR vs. ≈45 TUMOR FREE), making the raw accuracy score misleading
- AUC measures rank-ordering ability across all thresholds, not just at 0.5 cutoff
- It is the standard metric for binary classification in clinical/biomarker studies


```python
# Training/Validation/Test split evaluation
# Rule: fit on TRAIN, tune on VAL, report final performance on TEST (touched once only)
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score, RocCurveDisplay, ConfusionMatrixDisplay
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Load all three pre-split datasets (portable relative paths)
train_expr = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TRAINING_SET_GSE62944_subsample_log2TPM.csv', index_col=0)
val_expr   = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/VALIDATION_SET_GSE62944_subsample_log2TPM.csv', index_col=0)
test_expr  = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TEST_SET_GSE62944_subsample_log2TPM.csv', index_col=0)
train_meta = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TRAINING_SET_GSE62944_metadata.csv')
val_meta   = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/VALIDATION_SET_GSE62944_metadata.csv')
test_meta  = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TEST_SET_GSE62944_metadata.csv')

def prep_dataset(expr, meta, genes=SIG_GENES):
    # Extract LUAD samples with known tumor status.
    # Returns X (samples x features, DataFrame) and y (binary Series: 1=WITH TUMOR, 0=TUMOR FREE).
    
    m = meta[meta['cancer_type'] == 'LUAD'].copy()
    m['tsc'] = m['tumor_status'].where(m['tumor_status'].isin(['TUMOR FREE', 'WITH TUMOR']), other=np.nan)
    m = m.dropna(subset=['tsc'])
    # Keep only samples present in the expression matrix
    samps = [s for s in m['sample'].values if s in expr.columns]
    m = m[m['sample'].isin(samps)].set_index('sample')
    X = expr.loc[genes, samps].T.copy()   # samples x signature genes
    y = (m.loc[samps, 'tsc'] == 'WITH TUMOR').astype(int)
    return X, y

# Prepare datasets for supervised learning
X_train, y_train = prep_dataset(train_expr, train_meta)
X_val,   y_val   = prep_dataset(val_expr,   val_meta)
X_test,  y_test  = prep_dataset(test_expr,  test_meta)

# i used claude to string cacatenate
print(f'Train: n={len(y_train)}, WITH TUMOR={y_train.sum()}, TUMOR FREE={(y_train==0).sum()}')
print(f'Val:   n={len(y_val)},   WITH TUMOR={y_val.sum()},   TUMOR FREE={(y_val==0).sum()}')
print(f'Test:  n={len(y_test)},  WITH TUMOR={y_test.sum()},  TUMOR FREE={(y_test==0).sum()}')

# Standardize features — fit ONLY on training set, then apply same transform to val & test
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)  # learn mean/std from training data only
X_val_sc   = scaler.transform(X_val)         # apply training statistics to validation
X_test_sc  = scaler.transform(X_test)        # apply training statistics to test

# Random Forest (100 trees, unlimited depth)
rf_base = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',   # compensate for ~3:1 TUMOR FREE : WITH TUMOR imbalance
    random_state=42,
    max_features='sqrt'
)
rf_base.fit(X_train, y_train)
auc_train_base = roc_auc_score(y_train, rf_base.predict_proba(X_train)[:, 1])
auc_val_base   = roc_auc_score(y_val,   rf_base.predict_proba(X_val)[:, 1])

# i used claude to string catacentate 
print(f'\n=== BASELINE: Random Forest (100 trees) ===')
print(f'  Train AUC (in-sample):          {auc_train_base:.4f}')
print(f'  Validation AUC (out-of-sample): {auc_val_base:.4f}')
print(f'  Overfitting gap:                {auc_train_base - auc_val_base:.4f}')
print(f'  → Train AUC ≈ 1.00 but Val AUC << 1.00 confirms severe overfitting on this small cohort.')

# Logistic Regression with L2 regularization to reduce overfitting and improve generalization
# Motivation: RF memorizes individual patients on n≈56 training samples.
# L2-penalized logistic regression shrinks coefficients toward zero,
# capturing only the strongest, most consistent signal across patients.
lr_imp = LogisticRegression(
    C=0.1,               # strong L2 penalty (smaller C = more regularization)
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)
lr_imp.fit(X_train_sc, y_train)

# Evaluate improved model on training and validation sets
auc_train_imp = roc_auc_score(y_train, lr_imp.predict_proba(X_train_sc)[:, 1])
auc_val_imp   = roc_auc_score(y_val,   lr_imp.predict_proba(X_val_sc)[:, 1])
y_pred_val    = lr_imp.predict(X_val_sc)

# i used claude to string catacentate for clarity
print(f'\n=== IMPROVED: Logistic Regression (C=0.1, L2) ===')
print(f'  Train AUC (in-sample):          {auc_train_imp:.4f}')
print(f'  Validation AUC (out-of-sample): {auc_val_imp:.4f}')
print(f'  Overfitting gap:                {auc_train_imp - auc_val_imp:.4f}')
print(f'  ✓ Validation AUC change vs baseline: {auc_val_imp - auc_val_base:+.4f}')
print(f'  ✓ Overfitting gap reduced by:        {(auc_train_base - auc_val_base) - (auc_train_imp - auc_val_imp):.4f}')
print(f'\nValidation set classification report:')
print(classification_report(y_val, y_pred_val, target_names=['TUMOR FREE', 'WITH TUMOR']))

#FINAL TEST SET EVALUATION (touch only once)
auc_test_imp  = roc_auc_score(y_test, lr_imp.predict_proba(X_test_sc)[:, 1])
y_pred_test   = lr_imp.predict(X_test_sc)
y_prob_test   = lr_imp.predict_proba(X_test_sc)[:, 1]

# i used claude to string catacentate
print(f'\n=== FINAL TEST SET PERFORMANCE (Improved LR) ===')
print(f'  Test AUC:      {auc_test_imp:.4f}')
print(f'  Test Accuracy: {accuracy_score(y_test, y_pred_test):.4f}')
print(f'\nTest set classification report:')
print(classification_report(y_test, y_pred_test, target_names=['TUMOR FREE', 'WITH TUMOR']))

# Store probabilities for plotting in the next cell
y_prob_train_base = rf_base.predict_proba(X_train)[:, 1]
y_prob_val_base   = rf_base.predict_proba(X_val)[:, 1]
y_prob_train_imp  = lr_imp.predict_proba(X_train_sc)[:, 1]
y_prob_val_imp    = lr_imp.predict_proba(X_val_sc)[:, 1]
```

### Model Improvement: Switching to Logistic Regression with L2 Regularization

#### What We Changed
We replaced the baseline Random Forest (100 trees, unlimited depth, no regularization) with a *Logistic Regression using strong L2 regularization (C=0.1)* and standardized input features.

#### Why This Improves Generalization

The baseline Random Forest achieves *train AUC ≈ 1.000* but a substantially lower validation AUC, an overfitting gap of approximately *0.33*. On a training set of only ~56 samples, an unconstrained forest can grow trees deep enough to memorize individual patients. The result is a model that is highly accurate on training data but fails to generalize to new patients.

*Logistic Regression with L2 penalty* addresses this directly:
- The L2 penalty term adds a cost proportional to the *squared magnitude* of each coefficient, shrinking all coefficients toward zero
- This forces the model to rely only on the most consistent signal across patients, not on individual patient quirks
- *C=0.1* (a strong penalty) is appropriate here because our sample size is very small

#### Validation Performance Improvement

After switching to Logistic Regression, the *validation AUC increases* and the *overfitting gap decreases by approximately 0.15* (from ~0.33 to ~0.18). 

This means:
- The model generalizes better to unseen patients
- The gap between what the model "knows" about training patients vs. new patients has narrowed substantially
- The validation set, which the model never saw during training, is better classified

This validation AUC improvement is the key for model improvement: it tells us that the change was genuine and not just a reshuffling of training performance.

#### Feature Interpretation

The top predictive genes in the Logistic Regression (by absolute L2-regularized coefficient magnitude) are *HMMR* and *ANGPTL4*, the same genes that topped the Random Forest feature importances and that *Shen et al. #### Limitation

The validation AUC of approximately 0.68 is above random chance (0.50) but modest. ## Training & Evaluation: In-Sample vs. Out-of-Sample Error

To properly assess model generalization, we split the LUAD cohort into a *training set*  and a *validation set* using stratified splitting to preserve the class ratios. We fit the Random Forest on the training set only, computed *in-sample error* (performance on the training set), and then evaluated *out-of-sample error* (performance on the held-out validation set). This distinction was critical: as a model that is perfectly accurate on training data but poor on validation data is *overfit*, meaning it has memorized the training samples rather than learning generalizable patterns.

We use *AUC (Area Under the ROC Curve)* as our primary metric because:
- Our classes are imbalanced (≈17 WITH TUMOR vs. ≈45 TUMOR FREE), making the raw accuracy score misleading
- AUC measures rank-ordering ability across all thresholds, not just at 0.5 cutoff
- It is the standard metric for binary classification in clinical/biomarker studies



```python
# Training/Validation/Test split evaluation
# Rule: fit on TRAIN, tune on VAL, report final performance on TEST (touched once only)
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score, RocCurveDisplay, ConfusionMatrixDisplay
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Load all three pre-split datasets (portable relative paths)
train_expr = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TRAINING_SET_GSE62944_subsample_log2TPM.csv', index_col=0)
val_expr   = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/VALIDATION_SET_GSE62944_subsample_log2TPM.csv', index_col=0)
test_expr  = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TEST_SET_GSE62944_subsample_log2TPM.csv', index_col=0)
train_meta = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TRAINING_SET_GSE62944_metadata.csv')
val_meta   = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/VALIDATION_SET_GSE62944_metadata.csv')
test_meta  = pd.read_csv('/Users/isabelvikesland/Documents/UVA2/BME 2315 - Comp.BME/Module-4-Cancer/data/TEST_SET_GSE62944_metadata.csv')

def prep_dataset(expr, meta, genes=SIG_GENES):
    # Extract LUAD samples with known tumor status.
    # Returns X (samples x features, DataFrame) and y (binary Series: 1=WITH TUMOR, 0=TUMOR FREE).
    
    m = meta[meta['cancer_type'] == 'LUAD'].copy()
    m['tsc'] = m['tumor_status'].where(m['tumor_status'].isin(['TUMOR FREE', 'WITH TUMOR']), other=np.nan)
    m = m.dropna(subset=['tsc'])
    # Keep only samples present in the expression matrix
    samps = [s for s in m['sample'].values if s in expr.columns]
    m = m[m['sample'].isin(samps)].set_index('sample')
    X = expr.loc[genes, samps].T.copy()   # samples x signature genes
    y = (m.loc[samps, 'tsc'] == 'WITH TUMOR').astype(int)
    return X, y

# Prepare datasets for supervised learning
X_train, y_train = prep_dataset(train_expr, train_meta)
X_val,   y_val   = prep_dataset(val_expr,   val_meta)
X_test,  y_test  = prep_dataset(test_expr,  test_meta)

# i used claude to string cacatenate
print(f'Train: n={len(y_train)}, WITH TUMOR={y_train.sum()}, TUMOR FREE={(y_train==0).sum()}')
print(f'Val:   n={len(y_val)},   WITH TUMOR={y_val.sum()},   TUMOR FREE={(y_val==0).sum()}')
print(f'Test:  n={len(y_test)},  WITH TUMOR={y_test.sum()},  TUMOR FREE={(y_test==0).sum()}')

# Standardize features — fit ONLY on training set, then apply same transform to val & test
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)  # learn mean/std from training data only
X_val_sc   = scaler.transform(X_val)         # apply training statistics to validation
X_test_sc  = scaler.transform(X_test)        # apply training statistics to test

# Random Forest (100 trees, unlimited depth)
rf_base = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',   # compensate for ~3:1 TUMOR FREE : WITH TUMOR imbalance
    random_state=42,
    max_features='sqrt'
)
rf_base.fit(X_train, y_train)
auc_train_base = roc_auc_score(y_train, rf_base.predict_proba(X_train)[:, 1])
auc_val_base   = roc_auc_score(y_val,   rf_base.predict_proba(X_val)[:, 1])

# i used claude to string catacentate 
print(f'\n=== BASELINE: Random Forest (100 trees) ===')
print(f'  Train AUC (in-sample):          {auc_train_base:.4f}')
print(f'  Validation AUC (out-of-sample): {auc_val_base:.4f}')
print(f'  Overfitting gap:                {auc_train_base - auc_val_base:.4f}')
print(f'  → Train AUC ≈ 1.00 but Val AUC << 1.00 confirms severe overfitting on this small cohort.')

# Logistic Regression with L2 regularization to reduce overfitting and improve generalization
# Motivation: RF memorizes individual patients on n≈56 training samples.
# L2-penalized logistic regression shrinks coefficients toward zero,
# capturing only the strongest, most consistent signal across patients.
lr_imp = LogisticRegression(
    C=0.1,               # strong L2 penalty (smaller C = more regularization)
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)
lr_imp.fit(X_train_sc, y_train)

# Evaluate improved model on training and validation sets
auc_train_imp = roc_auc_score(y_train, lr_imp.predict_proba(X_train_sc)[:, 1])
auc_val_imp   = roc_auc_score(y_val,   lr_imp.predict_proba(X_val_sc)[:, 1])
y_pred_val    = lr_imp.predict(X_val_sc)

# i used claude to string catacentate for clarity
print(f'\n=== IMPROVED: Logistic Regression (C=0.1, L2) ===')
print(f'  Train AUC (in-sample):          {auc_train_imp:.4f}')
print(f'  Validation AUC (out-of-sample): {auc_val_imp:.4f}')
print(f'  Overfitting gap:                {auc_train_imp - auc_val_imp:.4f}')
print(f'  ✓ Validation AUC change vs baseline: {auc_val_imp - auc_val_base:+.4f}')
print(f'  ✓ Overfitting gap reduced by:        {(auc_train_base - auc_val_base) - (auc_train_imp - auc_val_imp):.4f}')
print(f'\nValidation set classification report:')
print(classification_report(y_val, y_pred_val, target_names=['TUMOR FREE', 'WITH TUMOR']))

#FINAL TEST SET EVALUATION (touch only once)
auc_test_imp  = roc_auc_score(y_test, lr_imp.predict_proba(X_test_sc)[:, 1])
y_pred_test   = lr_imp.predict(X_test_sc)
y_prob_test   = lr_imp.predict_proba(X_test_sc)[:, 1]

# i used claude to string catacentate
print(f'\n=== FINAL TEST SET PERFORMANCE (Improved LR) ===')
print(f'  Test AUC:      {auc_test_imp:.4f}')
print(f'  Test Accuracy: {accuracy_score(y_test, y_pred_test):.4f}')
print(f'\nTest set classification report:')
print(classification_report(y_test, y_pred_test, target_names=['TUMOR FREE', 'WITH TUMOR']))

# Store probabilities for plotting in the next cell
y_prob_train_base = rf_base.predict_proba(X_train)[:, 1]
y_prob_val_base   = rf_base.predict_proba(X_val)[:, 1]
y_prob_train_imp  = lr_imp.predict_proba(X_train_sc)[:, 1]
y_prob_val_imp    = lr_imp.predict_proba(X_val_sc)[:, 1]
```

### Model Improvement: Switching to Logistic Regression with L2 Regularization

#### What We Changed
We replaced the baseline Random Forest (100 trees, unlimited depth, no regularization) with a *Logistic Regression using strong L2 regularization (C=0.1)* and standardized input features.

#### Why This Improves Generalization

The baseline Random Forest achieves *train AUC ≈ 1.000* but a substantially lower validation AUC, an overfitting gap of approximately *0.33*. On a training set of only ~56 samples, an unconstrained forest can grow trees deep enough to memorize individual patients. The result is a model that is highly accurate on training data but fails to generalize to new patients.

*Logistic Regression with L2 penalty* addresses this directly:
- The L2 penalty term adds a cost proportional to the *squared magnitude* of each coefficient, shrinking all coefficients toward zero
- This forces the model to rely only on the most consistent signal across patients, not on individual patient quirks
- *C=0.1* (a strong penalty) is appropriate here because our sample size is very small

#### Validation Performance Improvement

After switching to Logistic Regression, the *validation AUC increases* and the *overfitting gap decreases by approximately 0.15* (from ~0.33 to ~0.18). 

This means:
- The model generalizes better to unseen patients
- The gap between what the model "knows" about training patients vs. new patients has narrowed substantially
- The validation set, which the model never saw during training, is better classified

This validation AUC improvement is the key for model improvement: it tells us that the change was genuine and not just a reshuffling of training performance.

#### Feature Interpretation

The top predictive genes in the Logistic Regression (by absolute L2-regularized coefficient magnitude) are *HMMR* and *ANGPTL4*, the same genes that topped the Random Forest feature importances and that *Shen et al. (2019)* independently identified as the strongest survival predictors. This convergence across two model types and an independent published study provides *external validation*: the signature is capturing real ECM invasion/metastasis biology, not just statistical noise.

#### Limitation

The validation AUC of approximately 0.68 is above random chance (0.50) but modest. This reflects: (1) a very small training cohort; (2) predicting *tumor status* rather than *survival*, the outcome the signature was originally validated for; and (3) the inherent biological noise in LUAD, a heterogeneous disease. These limitations are discussed further in the Conclusions section.

## Limitations and Future Work

### Limitations

*1. Small and imbalanced cohort.*
After subsetting to LUAD samples with complete tumor status labels from the GSE62944 dataset, we were left with approximately 62 usable patients (~17 WITH TUMOR, ~45 TUMOR FREE). This severe class imbalance and small n mean our validation set contains only ~19 samples, giving extremely wide confidence intervals around any AUC estimate. Conclusions about the signature's generalizability must therefore be interpreted cautiously.

*2. Surrogate outcome (tumor status) vs. true clinical endpoint (survival).*
Our original goal was to predict overall survival (OS) or disease-specific survival (DSS). We pivoted to binary tumor status classification because survival time data were sparse in the available subset. Tumor status is a weak proxy for survival: many patients with residual tumor live for years, while some tumor-free patients have already died of other causes. Conclusions about survival prediction cannot be directly drawn from this analysis.

*3. Dataset representativeness.*
The TCGA cohort disproportionately represents patients from large academic medical centers, predominantly of European ancestry. Gene expression patterns and clinical outcomes may differ in underrepresented populations (e.g., East Asian patients, where *EGFR* mutations are more prevalent). A model trained on TCGA data may not generalize equitably across all demographics.

*4. Only 9 gene features.*
We restricted features to the 9-gene ECM metastasis signature from Shen et al. (2019) to test a specific biological hypothesis. This constraint limits predictive power compared to whole-transcriptome approaches. The signature was validated for OS in a different cohort; its applicability to TCGA tumor status is an untested extrapolation.

*5. No external test set.*
All model evaluation was performed on held-out splits of the same GSE62944 dataset. True external validation on an entirely independent LUAD cohort (e.g., GEO dataset GSE31210 or CPTAC-LUAD) was not performed due to data access and scope constraints.

### Future Work

*1. Apply the signature to a larger, external LUAD cohort.*
Validating the 9-gene signature in an independent dataset (e.g., GSE31210 or the CPTAC LUAD proteogenomic cohort) with complete survival annotations would provide a genuine test of generalizability. This is the most important next step.

*2. Return to the original survival prediction goal.*
With a larger dataset, a Cox proportional-hazards model or survival random forest (using the lifelines or scikit-survival libraries) could directly assess whether the 9-gene signature is an independent predictor of DSS, adjusting for stage, age, and smoking history.

*3. Expand the feature set.*
Incorporating additional EMT-related genes (e.g., *VIM*, *CDH1*, *SNAI1*, *ZEB1*, *TWIST1*) or the full EMT gene signature from the MSigDB hallmark gene sets could improve predictive performance and provide a more complete picture of invasion-metastasis biology in LUAD.
